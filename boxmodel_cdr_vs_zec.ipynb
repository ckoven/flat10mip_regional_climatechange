```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
tau = 12. # years

s_0 = 0.06 # instantaneous annual sink fraction of last years emissions
```

```python
emissions_cdr = np.zeros(300)
emissions_cdr[0:100] = 10.
emissions_cdr[100:200] = 10. - np.arange(100)*0.2
emissions_cdr[200:300] = -10.
```

```python
emissions_zec = np.zeros(300)
emissions_zec[0:100] = 10.
emissions_zec[100:300] = 0.
```

```python
plt.plot(np.arange(300), emissions_cdr)
plt.plot(np.arange(300), emissions_zec, c='black')
```

```python
co2_pi = 280.
ntime = 300
co2_timeseries_cdr = np.zeros(ntime)/np.zeros(ntime)

co2_timeseries_cdr[0] = co2_pi
for t in range(1, ntime):
    sink_cdr = 0.
    for tt in range (0,t):
        sink_cdr = sink_cdr + s_0 * emissions_cdr[t-tt] * np.exp(-float(tt)/tau)
    co2_timeseries_cdr[t] = co2_timeseries_cdr[t-1] + emissions_cdr[t-1] - sink_cdr
        
```

```python
co2_pi = 280.
ntime = 300
co2_timeseries_zec = np.zeros(ntime)/np.zeros(ntime)

co2_timeseries_zec[0] = co2_pi
for t in range(1, ntime):
    sink_zec = 0.
    for tt in range (0,t):
        sink_zec = sink_zec + s_0 * emissions_zec[t-tt] * np.exp(-float(tt)/tau)
    co2_timeseries_zec[t] = co2_timeseries_zec[t-1] + emissions_zec[t-1] - sink_zec
        
```

```python
plt.plot(np.arange(300), co2_timeseries_cdr, label='esm-flat10-cdr')
plt.plot(np.arange(300), co2_timeseries_zec, c='black', label='esm-flat10-zec')

plt.hlines(co2_pi, xmin=0,xmax=300)
plt.vlines((100.,200), ymin=200,ymax=700)

plt.ylabel('CO2 concentration')
plt.legend()
```
